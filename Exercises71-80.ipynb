## Exercise 71:

Given an array as a list, convert it to float with reversed order.

```python
ah = "2 4 6 -3 9"
arr = ah.strip().split(" ")
arr
```

```python
import numpy as np

def arrays(arr):
    a = np.array(arr, float)
    a = np.flip(a)
    return a

arrays(arr)
```

# Exercise 73:

Calculate mean, var and std of a given array along a specified axis.

```python
import numpy as np
arr = np.array([[1,2,3],[3,2,6]])

print(np.mean(arr, axis=1))
print(np.var(arr, axis=0))
print(round(np.std(arr, axis=None),3))
```

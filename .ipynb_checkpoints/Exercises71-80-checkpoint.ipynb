## Exercise 71:

Given an array as a list, convert it to float with reversed order.

```python
ah = "2 4 6 -3 9"
arr = ah.strip().split(" ")
arr
```

```python
import numpy as np

def arrays(arr):
    a = np.array(arr, float)
    a = np.flip(a)
    return a

arrays(arr)
```

# Exercise 72:
Reshape a given 1D array into (3,3) array.

```python
s = "1 2 3 4 5 6 7 8 9"
import numpy as np
s = s.split(" ")
s = np.array(s, int)
a = np.reshape(s,(3,3))
a
```
